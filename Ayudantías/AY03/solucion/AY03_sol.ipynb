## 🧬 Resumen esencial de herencia

Una clase hija **hereda** todos los atributos y métodos de su clase padre, y puede agregar los suyos o **sobreescribir** los que ya existen. `super()` sirve para llamar a la versión de la clase padre sin repetir su código.

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:----------------------|:------------|
| **Heredar** | `class Hija(Padre):` | La clase hija recibe todos los atributos y métodos de la clase padre | `class Arquero(Jugador):` |
| **Llamar a la clase padre** | `super().__init__(args)` | Ejecuta el inicializador de la clase padre desde la clase hija | `super().__init__(apellido, numero, equipo)` |
| **Extender el `__init__`** | `super().__init__(...)` y luego los atributos propios | Primero lo de la clase padre, después lo propio | `self.porc_atajadas = porc_atajadas` |
| **Sobreescribir (*override*)** | Definir un método con el **mismo nombre** que el de la clase padre | La versión de la clase hija reemplaza a la de la clase padre | `def pasar(self, jugador_2): ...` |
| **Extender un método** | `super().metodo(args)` dentro del *override* | Reutiliza lo que hace la clase padre y le agrega algo | `super().pasar(j2)` y después un `print` |
| **¿Es de esta clase?** | `isinstance(obj, Clase)` | `True` también si `obj` es de una subclase | `isinstance(arquero, Jugador)` |
| **¿Hereda de?** | `issubclass(Hija, Padre)` | Revisa la relación entre dos clases | `issubclass(Arquero, Jugador)` |

> 💡 La pregunta que conviene hacerse es **"¿es un?"**: un `Arquero` **es un** `Jugador`, así que todo lo que sabe hacer `Jugador` no hay que escribirlo de nuevo.

```python
from random import random


class Jugador:

    def __init__(self, apellido, numero, equipo):
        self.apellido = apellido
        self.numero = numero
        self.equipo = equipo
        self.tiene_balon = False
    
    def pasar(self, jugador_2):
        print(f"{self.apellido} se la pasa a {jugador_2.apellido}")
        self.tiene_balon = False
        jugador_2.tiene_balon = True


class Arquero(Jugador):

    def __init__(self, apellido, numero, equipo, porc_atajadas):
        super().__init__(apellido, numero, equipo)
        self.porc_atajadas = porc_atajadas
    
    def atajar(self):
        self.tiene_balon = True
        if random() <= self.porc_atajadas:
            print(f"Atajadón de {self.apellido}!!! Hay portero señores!!!!")
            return True
        else:
            return False


class Defensa(Jugador):

    def __init__(self, apellido, numero, equipo, porc_barridas):
        super().__init__(apellido, numero, equipo)
        self.porc_barridas = porc_barridas
    
    def barrerse(self, rival):
        if random() <= self.porc_barridas:
            print(f"{self.apellido} le quita el balón a {rival.apellido}")
            rival.tiene_balon = False
            self.tiene_balon = True
        else:
            print(f"{self.apellido} le trata de quitar el balón a {rival.apellido}... pero no puede!")


class Delantero(Jugador):

    def __init__(self, apellido, numero, equipo, porc_tiros):
        super().__init__(apellido, numero, equipo)
        self.porc_tiros = porc_tiros
    
    def tirar(self, arquero_rival):
        if random() <= self.porc_tiros:
            print(f"{self.apellido} chuta...")
            if not arquero_rival.atajar():
                print(f"GOLAZO DEL {self.equipo.upper()}!!! Un tiro al ángulo de {self.apellido}!!")
        else:
            print(f"{self.apellido} chuta... y fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True


colo_coloop = [Arquero("Bravo", 1, "Colo-ColOOP", 0.55),
               Defensa("Riffo", 4, "Colo-ColOOP", 0.55),
               Defensa("Henríquez", 5, "Colo-ColOOP", 0.40),
               Delantero("Suazo", 11, "Colo-ColOOP", 0.95),
               Delantero("Vidal", 22, "Colo-ColOOP", 0.8)]

codebreloa = [Arquero("Tapia", 1, "Codebreloa", 0.3),
              Defensa("Fuentes", 5, "Codebreloa", 0.45),
              Defensa("González", 3, "Codebreloa", 0.35),
              Delantero("Norambuena", 9, "Codebreloa", 0.95),
              Delantero("Galaz", 7, "Codebreloa", 0.8)]

colo_coloop[0].tiene_balon = True
colo_coloop[0].pasar(colo_coloop[1])
colo_coloop[1].pasar(colo_coloop[4])
codebreloa[1].barrerse(colo_coloop[4])
if codebreloa[1].tiene_balon:
    codebreloa[1].pasar(codebreloa[4])
    colo_coloop[2].barrerse(codebreloa[4])
    if codebreloa[4].tiene_balon:
        codebreloa[4].pasar(codebreloa[3])
        codebreloa[3].tirar(colo_coloop[0])
else:
    colo_coloop[4].pasar(colo_coloop[3])
    colo_coloop[3].tirar(codebreloa[0])
```

## 🧩 Resumen esencial de multiherencia

Una clase puede heredar de **varias** clases a la vez. Python resuelve de dónde sale cada método usando el **MRO** (*Method Resolution Order*): recorre las clases padre de izquierda a derecha, y nunca pone una clase antes que alguna de sus clases hijas.

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:----------------------|:------------|
| **Heredar de varias** | `class Hija(A, B):` | El orden importa: `A` se revisa antes que `B` | `class Lateral(Defensa, Delantero):` |
| **Ver el orden** | `Clase.mro()` o `Clase.__mro__` | Lista el orden exacto en que se buscan los métodos | `Lateral.mro()` |
| **Un solo `super()`** | `super().__init__(**kwargs)` | Recorre **toda** la cadena del MRO, pasando una vez por cada clase | `super().__init__(**kwargs)` |
| **Pasar argumentos hacia arriba** | `def __init__(self, propio, **kwargs)` | Cada clase se queda con lo suyo y manda el resto al siguiente del MRO | `def __init__(self, porc_barridas, **kwargs):` |
| **Llamar a una clase puntual** | `Clase.metodo(self, args)` | Se salta el MRO y usa esa implementación directamente | `Defensa.barrerse(self, rival)` |
| **Método repetido en A y B** | Gana el primero del MRO | Si `A` y `B` definen `metodo`, se usa el de `A` | `Lateral(Defensa, Delantero)` usa el de `Defensa` |
| **Problema del diamante** | Dos caminos llegan a la misma clase base | Con llamadas directas la base se ejecuta **dos veces**; con `super()`, una sola | `Anfibio(Terrestre, Acuatico)` |

```python
class Defensa(Jugador):

    def __init__(self, porc_barridas, **kwargs):
        super().__init__(**kwargs)
        self.porc_barridas = porc_barridas
    
    def barrerse(self, rival):
        if random() <= self.porc_barridas:
            print(f"{self.apellido} le quita el balón a {rival.apellido}")
            rival.tiene_balon = False
            self.tiene_balon = True
        else:
            print(f"{self.apellido} le trata de quitar el balón a {rival.apellido}... pero no puede!")


class Delantero(Jugador):

    def __init__(self, porc_tiros, porc_cabeceo, **kwargs):
        super().__init__(**kwargs)
        self.porc_tiros = porc_tiros
        self.porc_cabeceo = porc_cabeceo
    
    def tirar(self, arquero_rival):
        if random() <= self.porc_tiros:
            print(f"{self.apellido} chuta...")
            if not arquero_rival.atajar():
                print(f"GOLAZO DEL {self.equipo.upper()}!!! Un tiro al ángulo de {self.apellido}!!")
        else:
            print(f"{self.apellido} chuta... y fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True
    
    def cabecear(self, arquero_rival):
        if random() <= self.porc_cabeceo:
            print(f"{self.apellido} cabecea...")
            if not arquero_rival.atajar():
                print(f"GOLAZO DEL {self.equipo.upper()}!!! Un cabezazo fulminante de {self.apellido}!!")
        else:
            print(f"{self.apellido} cabecea... y fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True


class Lateral(Defensa, Delantero):

    def __init__(self, porc_centros, **kwargs):
        super().__init__(**kwargs)
        self.porc_centros = porc_centros
    
    def centrar(self, delantero, arquero_rival):
        if random() <= self.porc_centros:
            print(f"Centro de {self.apellido} para {delantero.apellido}")
            self.tiene_balon = False
            delantero.tiene_balon = True
            delantero.cabecear(arquero_rival)
        else:
            print(f"Centro de {self.apellido}... se va fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True


colo_coloop = [Arquero(apellido="Bravo", numero=1, equipo="Colo-ColOOP", porc_atajadas=0.55),
              Defensa(apellido="Riffo", numero=4, equipo="Colo-ColOOP", porc_barridas=0.55),
              Defensa(apellido="Henríquez", numero=5, equipo="Colo-ColOOP", porc_barridas=0.4),
              Lateral(apellido="Valdivia", numero=6, equipo="Colo-ColOOP", porc_barridas=0.4, porc_cabeceo=0.6, porc_tiros=0.7, porc_centros=0.75),
              Delantero(apellido="Suazo", numero=11, equipo="Colo-ColOOP", porc_tiros=0.95, porc_cabeceo=0.8),
              Delantero(apellido="Vidal", numero=22, equipo="Colo-ColOOP", porc_tiros=0.8, porc_cabeceo=0.8)]

codebreloa = [Arquero(apellido="Tapia", numero=1, equipo="Codebreloa", porc_atajadas=0.3),
                   Defensa(apellido="Fuentes", numero=5, equipo="Codebreloa", porc_barridas=0.45),
                   Defensa(apellido="González", numero=3, equipo="Codebreloa", porc_barridas=0.35),
                   Lateral(apellido="Sánchez", numero=24, equipo="Codebreloa", porc_barridas=0.3, porc_cabeceo=0.5, porc_tiros=0.8, porc_centros=0.6),
                   Delantero(apellido="Norambuena", numero=9, equipo="Codebreloa", porc_tiros=0.95, porc_cabeceo=0.99),
                   Delantero(apellido="Galaz", numero=7, equipo="Codebreloa", porc_tiros=0.8, porc_cabeceo=0.9)]

colo_coloop[0].tiene_balon = True
colo_coloop[0].pasar(colo_coloop[1])
colo_coloop[1].pasar(colo_coloop[3])
codebreloa[3].barrerse(colo_coloop[3])
if codebreloa[3].tiene_balon:
    colo_coloop[2].barrerse(codebreloa[3])
    if codebreloa[3].tiene_balon:
        codebreloa[3].tirar(colo_coloop[0])
else:
    colo_coloop[3].centrar(colo_coloop[5], codebreloa[0])
```

## 🔒 Resumen esencial de *properties*

Una *property* es un atributo que por detrás son métodos: se lee y se escribe como `objeto.atributo`, pero cada acceso pasa por código que nosotros controlamos. Sirve para validar, transformar o avisar cada vez que un valor se lee o se modifica.

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:----------------------|:------------|
| **Atributo interno** | `self._x = valor` | Por convención el `_` marca "esto no se toca desde afuera" | `self._porc_atajadas = 0.55` |
| ***Getter*** | `@property` sobre `def x(self)` | Se ejecuta al **leer** `objeto.x`; debe **retornar** el valor | `@property` sobre `def porc_atajadas(self)` |
| ***Setter*** | `@x.setter` sobre `def x(self, valor)` | Se ejecuta al **asignar** `objeto.x = valor` | `@porc_atajadas.setter` |
| **Leer** | `objeto.x` | Sin paréntesis: por fuera se ve igual que un atributo | `arquero.porc_atajadas` |
| **Escribir** | `objeto.x = valor` | Dispara el *setter*; no reemplaza la *property* | `arquero.porc_atajadas += 0.05` |
| **Solo lectura** | Definir el *getter* y **no** el *setter* | Asignar lanza `AttributeError` | `@property` sobre `def area(self)` |
| **Forma funcional** | `x = property(fget, fset, fdel)` | Lo mismo, pero sin decoradores | `email = property(_get_email, _set_email)` |

> ⚠️ Dentro del *getter* y del *setter* hay que usar el atributo interno (`self._x`), **nunca** `self.x`: eso volvería a llamar a la *property* y caería en una recursión infinita.

```python
class Arquero(Jugador):

    def __init__(self, apellido, numero, equipo, porc_atajadas):
        super().__init__(apellido, numero, equipo)
        self.porc_atajadas_ = porc_atajadas
    
    @property
    def porc_atajadas(self):
        return self.porc_atajadas_

    @porc_atajadas.setter
    def porc_atajadas(self, nuevo_valor):
        if nuevo_valor <= 0.7:
            self.porc_atajadas_ = round(nuevo_valor, 2)
            print(f"Debido a ese tiro tan malo, {self.apellido} sube su porcentaje de atajadas a {self.porc_atajadas}")
        else:
            print(f"{self.apellido} no puede subir más su porcentaje de atajadas!")
        
    
    def atajar(self):
        self.tiene_balon = True
        if random() <= self.porc_atajadas:
            print(f"Atajadón de {self.apellido}!!! Hay portero señores!!!!")
            return True
        else:
            return False


class Defensa(Jugador):

    def __init__(self, porc_barridas, **kwargs):
        super().__init__(**kwargs)
        self.porc_barridas = porc_barridas
    
    def barrerse(self, rival):
        if random() <= self.porc_barridas:
            print(f"{self.apellido} le quita el balón a {rival.apellido}")
            rival.tiene_balon = False
            self.tiene_balon = True
        else:
            print(f"{self.apellido} le trata de quitar el balón a {rival.apellido}... pero no puede!")


class Delantero(Jugador):

    def __init__(self, porc_tiros, porc_cabeceo, **kwargs):
        super().__init__(**kwargs)
        self.porc_tiros_ = porc_tiros
        self.porc_cabeceo_ = porc_cabeceo
        self.on_fire = False

    @property
    def porc_tiros(self):
        if self.on_fire:
            print(f"{self.apellido} está on fire!!")
            return self.porc_tiros_ + 0.2
        else:
            return self.porc_tiros_
    
    @property
    def porc_cabeceo(self):
        if self.on_fire:
            print(f"{self.apellido} está on fire!!")
            return self.porc_cabeceo_ + 0.2
        else:
            return self.porc_cabeceo_
    
    
    def tirar(self, arquero_rival):
        if random() <= self.porc_tiros:
            print(f"{self.apellido} chuta...")
            if not arquero_rival.atajar():
                print(f"GOLAZO DEL {self.equipo.upper()}!!! Un tiro al ángulo de {self.apellido}!!")
        else:
            print(f"{self.apellido} chuta... y fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True
            arquero_rival.porc_atajadas = arquero_rival.porc_atajadas + 0.05
    
    def cabecear(self, arquero_rival):
        if random() <= self.porc_cabeceo:
            print(f"{self.apellido} cabecea...")
            if not arquero_rival.atajar():
                print(f"GOLAZO DEL {self.equipo.upper()}!!! Un cabezazo fulminante de {self.apellido}!!")
        else:
            print(f"{self.apellido} cabecea... y fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True

class Lateral(Defensa, Delantero):

    def __init__(self, porc_centros, **kwargs):
        super().__init__(**kwargs)
        self.porc_centros = porc_centros
    
    def centrar(self, delantero, arquero_rival):
        if random() <= self.porc_centros:
            print(f"Centro de {self.apellido} para {delantero.apellido}")
            self.tiene_balon = False
            delantero.tiene_balon = True
            delantero.cabecear(arquero_rival)
        else:
            print(f"Centro de {self.apellido}... se va fuera!")
            self.tiene_balon = False
            arquero_rival.tiene_balon = True


colo_coloop = [Arquero(apellido="Bravo", numero=1, equipo="Colo-ColOOP", porc_atajadas=0.35),
              Defensa(apellido="Riffo", numero=4, equipo="Colo-ColOOP", porc_barridas=0.55),
              Defensa(apellido="Henríquez", numero=5, equipo="Colo-ColOOP", porc_barridas=0.4),
              Lateral(apellido="Valdivia", numero=6, equipo="Colo-ColOOP", porc_barridas=0.4, porc_cabeceo=0.6, porc_tiros=0.7, porc_centros=0.75),
              Delantero(apellido="Suazo", numero=11, equipo="Colo-ColOOP", porc_tiros=0.95, porc_cabeceo=0.8),
              Delantero(apellido="Vidal", numero=22, equipo="Colo-ColOOP", porc_tiros=0.8, porc_cabeceo=0.8)]

codebreloa = [Arquero(apellido="Tapia", numero=1, equipo="Codebreloa", porc_atajadas=0.3),
                   Defensa(apellido="Fuentes", numero=5, equipo="Codebreloa", porc_barridas=0.45),
                   Defensa(apellido="González", numero=3, equipo="Codebreloa", porc_barridas=0.35),
                   Lateral(apellido="Sánchez", numero=24, equipo="Codebreloa", porc_barridas=0.3, porc_cabeceo=0.5, porc_tiros=0.3, porc_centros=0.6),
                   Delantero(apellido="Norambuena", numero=9, equipo="Codebreloa", porc_tiros=0.95, porc_cabeceo=0.99),
                   Delantero(apellido="Galaz", numero=7, equipo="Codebreloa", porc_tiros=0.8, porc_cabeceo=0.9)]


codebreloa[0].tiene_balon = True
codebreloa[0].pasar(codebreloa[3])
codebreloa[3].tirar(colo_coloop[0])
colo_coloop[0].tiene_balon = False
codebreloa[0].tiene_balon = True
codebreloa[0].pasar(codebreloa[3])
codebreloa[3].tirar(colo_coloop[0])
colo_coloop[0].tiene_balon = False
codebreloa[0].tiene_balon = True
codebreloa[0].pasar(codebreloa[3])
codebreloa[3].on_fire = True
codebreloa[3].tirar(colo_coloop[0])
colo_coloop[0].tiene_balon = False
codebreloa[0].tiene_balon = True
codebreloa[0].pasar(codebreloa[3])
codebreloa[3].tirar(colo_coloop[0])
colo_coloop[0].tiene_balon = False
codebreloa[0].tiene_balon = True
codebreloa[0].pasar(codebreloa[3])
codebreloa[3].on_fire = False
codebreloa[3].tirar(colo_coloop[0])
print(colo_coloop[0].porc_atajadas)
```
