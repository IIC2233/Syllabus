# Programación Funcional

## Paradigma Funcional

Python permite escribir código funcional, combinándolo con otros paradigmas. En este enfoque pensamos el programa como una **transformación de datos**: cada función recibe un estado y devuelve otro.

Conceptos centrales:
- **Funciones puras**: para el mismo input, entregan el mismo output.
- **Sin efectos secundarios**: la función no modifica estado externo (variables globales, archivos, consola, etc.).
- **Inmutabilidad lógica**: en vez de mutar estructuras, se construyen nuevas versiones.
- **Funciones de primera clase**: las funciones se pueden guardar en variables, pasar como argumento y retornar desde otras funciones.

Cuando una función es pura, es más fácil de razonar, reutilizar y optimizar (por ejemplo, con `@cache`).


## Funciones Generadoras

Una **función generadora** es una función que contiene `yield`.

Diferencias con una función tradicional:
- `return` termina la función inmediatamente.
- `yield` entrega un valor y **pausa** la ejecución, guardando su estado interno para continuar después.

Esto permite producir datos de forma incremental, evitando cargar todo en memoria a la vez.

Ideas clave:
- Al llamar la función, se obtiene un **generador** (no se ejecuta todo inmediatamente).
- Se puede recorrer con `for` o con `next()`.
- Un generador se consume una vez.
- `yield from` delega la iteración a otro iterable/generador.


`````python
def pares_hasta(n: int):
    for i in range(n + 1):
        if i % 2 == 0:
            yield i

gen = pares_hasta(10)
print(next(gen))
print(next(gen))
print(list(gen))  # continúa desde donde quedó

def generador_hijo():
    yield from range(3)

def generador_padre():
    yield -1
    yield from generador_hijo()
    yield 99

print(list(generador_padre()))
`````

## Funciones que trabajan con iterables

### Que necesitan de una función
- `map(f, iterable)`: aplica `f` a cada elemento.
- `filter(f, iterable)`: conserva elementos donde `f(elem)` es `True`.
- `reduce(f, iterable[, inicial])`: acumula el iterable a un solo resultado.

#### Funciones Lambda o funciones anónimas

Una forma alternativa de definir funciones que nos serán de utilidad a la hora de trabajar con `map`, `filter` y `reduce`. Características:
- Son definidas de forma anónima, es decir, no poseen un nombre específico.
- Son fugaces, lo que quiere decir que son utilizadas donde son definidas (se considera mala práctica guardar una función `lambda` en una variable).

`lambda <parámetros> : <valor a retornar>`

### Que no necesitan de una función

Funciones que encajan muy bien con el paradigma funcional porque operan directamente sobre iterables.

| Función       | Modo de uso                                         | Módulo        | Explicación breve                                                                 |
|--------------|------------------------------------------------------|---------------|-----------------------------------------------------------------------------------|
| `sum`        | `sum(iterable, start=0)`                             | `builtins`    | Suma los elementos de un iterable, opcionalmente partiendo de `start`.           |
| `min`        | `min(iterable)`               | `builtins`    | Devuelve el valor mínimo de un iterable o de varios argumentos.                  |
| `max`        | `max(iterable)`               | `builtins`    | Devuelve el valor máximo de un iterable o de varios argumentos.                  |
| `all`        | `all(iterable)`                                     | `builtins`    | Devuelve `True` si **todos** los elementos del iterable son verdaderos.          |
| `any`        | `any(iterable)`                                     | `builtins`    | Devuelve `True` si **algún** elemento del iterable es verdadero.                 |
| `enumerate`  | `enumerate(iterable, start=0)`                      | `builtins`    | Devuelve pares `(índice, valor)` al iterar sobre la secuencia.                   |
| `zip`        | `zip(iter1, iter2, ...)`                            | `builtins`    | Itera en paralelo, devolviendo tuplas con el i-ésimo elemento de cada iterable.  |
| `reversed`   | `reversed(seq)`                                     | `builtins`    | Devuelve un iterador que recorre la secuencia en orden inverso.                  |
| `count`      | `count(start=0, step=1)` | `itertools` | Genera una secuencia infinita: `start`, `start+step`, `start+2*step`, ...        |

| `repeat`     | `repeat(obj, times)` | `itertools`   | Devuelve el mismo objeto una y otra vez, opcionalmente un número fijo de veces.  |

`````python
from functools import reduce

numeros = [1, 2, 3, 4, 5, 6]

# Necesitan una función
cuadrados = list(map(lambda x: x ** 2, numeros))
pares = list(filter(lambda x: x % 2 == 0, numeros))
suma_total = reduce(lambda acc, x: acc + x, numeros, 0)

print("Cuadrados:", cuadrados)
print("Pares:", pares)
print("Suma total:", suma_total)


# No necesitan una función
print("\nEnumerate:")
for i, valor in enumerate(["a", "b", "c"], start=1):
    print(i, valor)

print("\nZip:")
nombres = ["Ana", "Luis", "Mora"]
edades = [21, 20, 23]
for par in zip(nombres, edades):
    print(par)

print("\nall/any:")
print(all(x > 0 for x in numeros))
print(any(x > 5 for x in numeros))
`````

# Threading

Muchas veces al programar necesitaremos hacer dos cosas al mismo tiempo, y para eso podemos usar la librería `threading`. Esta nos permite programar *concurrencia*. Es importante conocer la diferencia entre:


- **Concurrencia**: gestionar varias tareas de forma intercalada en un mismo núcleo o procesador. Aunque solo una tarea se ejecute en un instante dado, los cambios de contexto se realizan tan rápido que aparentan simultaneidad. Se usa la librería `threading`.
- **Paralelismo**: ejecución simultánea de varias tareas en diferentes núcleos o procesadores. Se usan otras librerías.

## Thread
Un `Thread` es un objeto que permite ejecutar un trozo de código de manera concurrente al hilo principal. Podemos definirlos con una función objetivo o hacer una clase que herede de `Thread` y sobreescriba el método `run`.

`````python
from threading import Thread
import random, time

def contar(n, paridad):

    for i in range(n):
        if i % 2 == paridad:
            time.sleep(random.randint(0, 1))
            print(i)


thread_1 = Thread(target=contar, args=(10, 0))
thread_2 = Thread(target=contar, args=(10, 1))

thread_1.start()
thread_2.start()

# al parecer algunos números se imprimen en la misma línea, cómo arreglamos eso? 🤔
`````

## join
Si estamos en la `thread_1` y ejecutamos `thread_2.join()`, vamos a parar la ejecución de la `thread_1` hasta que termine `thread_2`.

`````python
class MiThread(Thread):

    def __init__(self, numero):
        super().__init__()
        self.numero = numero

    def guardar(self, otra_thread):
        self.otra_thread = otra_thread
        
    def run(self):
        print(f"Thread {self.numero}: Voy a contar!")
        for i in range(10):
            print(f"Thread {self.numero}: {i}")
            time.sleep(0.5 * self.numero)

            # la thread 1 espera a la thread 2
            if i == 5 and self.numero == 1:
                print(f"Thread {self.numero}: esperaré a la otra thread para seguir")
                self.otra_thread.join()
        print(f"Thread {self.numero}: Terminé :v")


thread_1 = MiThread(1)
thread_2 = MiThread(2)
thread_1.guardar(thread_2)
thread_2.guardar(thread_1)

thread_1.start()
thread_2.start()
        
`````

## Daemon Thread
Una thread **daemon** es una thread que acaba su ejecución si es que la thread principal termina. Por defecto, las threads no son **daemon**, por lo que tenemos que esperar que terminen para terminar el programa principal.

## Lock

Cuando queremos que una única thread esté en una sección de código a la vez debemos usar la clase `Lock`. Esto nos sirve cuando tenemos acceso a recursos compartidos. Debemos usar `lock.acquire()` para adquirir un lock y `lock.release()` para soltarlo.

`````python
def sumador(lista):
    for i in range(10**6):
        lista[0] += int(1)
        # pregunta: por qué ponemos este int?

lista = [0]

thread_1 = Thread(target=sumador, args=(lista, ))
thread_2 = Thread(target=sumador, args=(lista, ))

thread_1.start()
thread_2.start() 

thread_1.join()
thread_2.join() 

print(lista)
# imprime cada vez algo distinto, watafak?¡
`````

`````python
from threading import Lock

def sumador(lista, lock):
    for i in range(10**6):
        lock.acquire()
        # en esta parte solo puede estar una línea a la vez
        lista[0] += int(1)
        lock.release()

lista = [0]
lock = Lock()

thread_1 = Thread(target=sumador, args=(lista, lock))
thread_2 = Thread(target=sumador, args=(lista, lock))

thread_1.start()
thread_2.start() 

thread_1.join()
thread_2.join() 

print(lista)
# ahora si :)
`````

## Events
Con la clase **Event** podemos hace que una thread espere a que pase algún evento en otra thread, sin necesidad de que la otra thread termine.

`````python
from threading import Event

class MiThread(Thread):

    def __init__(self, numero, evento):
        super().__init__()
        self.numero = numero
        self.evento = evento
        
    def run(self):
        print(f"Thread {self.numero}: Voy a contar!")

        if self.numero == 1:
            # la thread 1 espera a que la thread 2 cuente hasta 5
            self.evento.wait()

        for i in range(10):
            print(f"Thread {self.numero}: {i}")
            time.sleep(0.5 * self.numero)
            # cuando la thread 2 cuenta hasta 5, setea el evento
            if self.numero == 2 and i == 5:
                self.evento.set()
        print(f"Thread {self.numero}: Terminé :v")


evento = Event()
thread_1 = MiThread(1, evento)
thread_2 = MiThread(2, evento)

thread_1.start()
thread_2.start()
        
`````

## Deadlocks
Los **deadlocks** se producen cuando ambas threads están esperando a que la otra haga algo, por lo que ninguna puede avanzar.

`````python
from threading import Event

class MiThread(Thread):

    def __init__(self, numero, evento_1, evento_2):
        super().__init__()
        self.numero = numero
        self.evento_1 = evento_1
        self.evento_2 = evento_2
        
    def run(self):
        print(f"Thread {self.numero}: Voy a contar!")

        # la thread 1 espera al evento 2 para setear el evento 1
        if self.numero == 1:
            self.evento_2.wait()
            self.evento_1.set()

        # la thread 2 espera al evento 1 para setear el evento 2
        if self.numero == 2:
            self.evento_1.wait()
            self.evento_2.set()

        for i in range(10):
            print(f"Thread {self.numero}: {i}")
            time.sleep(0.5 * self.numero)
        print(f"Thread {self.numero}: Terminé :v")


evento_1 = Event()
evento_2 = Event()
thread_1 = MiThread(1, evento_1, evento_2)
thread_2 = MiThread(2, evento_1, evento_2)

thread_1.start()
thread_2.start()
        
`````

## Serialización y JSON

Cuando enviamos datos a través de la red, estos deben ser **serializados**. El formato que utilizaremos en este curso es **JSON (JavaScript Object Notation)**.

### ¿Qué es JSON?
JSON es un formato **multilenguaje**, es decir, puede ser leído y generado por prácticamente cualquier lenguaje: Python, Java, C++, JavaScript, Go, etc.
Además de esto es *human-readable*, es decir, puedes leer un archivo .json y entender lo que aparece en este.

### ¿Qué tipos de Python se pueden convertir a JSON?

| Tipo en Python | Representación JSON | ¿Es válido convertirlo? |
|----------------|---------------------|--------------------------|
| `dict`         | objeto JSON         | ✔ Sí |
| `list`, `tuple`| arreglo JSON        | ✔ Sí |
| `str`          | string              | ✔ Sí |
| `int`          | number              | ✔ Sí |
| `float`        | number              | ✔ Sí |
| `bool`         | true/false          | ✔ Sí |
| `None`         | null                | ✔ Sí |

### ¿Qué **no** se puede transformar directamente a JSON?
JSON **no admite** tipos complejos propios de Python, como:
- instancias de clases
- funciones
- sets (`set`)
- bytes
- datetime u otros objetos no estándar

Para enviar esos tipos, primero deben convertirse a valores compatibles: por ejemplo, `datetime → string`, `set → list`, `objeto → dict`.

### Ejemplo de JSON válido
```json

  "roles": ["admin", "usuario"],
  "activo": true
```

### Serialización de objetos personalizados

¿Qué pasa si quiero serializar una instancia de una clase? Como JSON no sabe hacerlo solo, debemos adaptar la información. Una opción rápida es serializar el `__dict__` del objeto, pero eso no nos permite recuperar la instancia original después.

Para personalizar bien la serialización y la deserialización usamos:
- `json.JSONEncoder` (sobrescribiendo el método `default`) para **serializar**.
- el parámetro `object_hook` para **deserializar**.

`````python
import json


class Mascota:
    def __init__(self, nombre, edad, especie):
        self.nombre = nombre
        self.edad = edad
        self.especie = especie

    def __repr__(self):
        return f"Mascota({self.nombre}, {self.edad}, {self.especie})"


# Para serializar: heredamos de JSONEncoder y sobrescribimos default
class MascotaEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, Mascota):
            return {"nombre": obj.nombre, "edad": obj.edad, "especie": obj.especie}
        # Para otros tipos, dejamos el comportamiento por defecto
        return super().default(obj)


luna = Mascota("Luna", 8, "Gato")

texto = json.dumps(luna, cls=MascotaEncoder)
print(texto)
`````

`````python
# Para deserializar: entregamos una función al parámetro object_hook.
# Cada objeto JSON (que llega como dict) pasa por esta función.
def mascota_hook(diccionario):
    return Mascota(diccionario["nombre"], diccionario["edad"], diccionario["especie"])


recuperada = json.loads(texto, object_hook=mascota_hook)
print(recuperada, type(recuperada))
`````

## Networking: conceptos básicos

Para que dos computadores se comuniquen por la red necesitamos identificarlos y ponernos de acuerdo en cómo hablar.

- **Dirección IP**: identifica a un dispositivo dentro de la red (por ejemplo `192.168.0.10`).
- **Puerto**: identifica un proceso o servicio dentro de ese dispositivo (por ejemplo el `5000`). La combinación IP + puerto define el extremo de una comunicación.

Sobre cómo se transmiten los datos, los dos protocolos principales son:

- **TCP** (*Transmission Control Protocol*): orientado a la conexión. Garantiza que los datos lleguen completos y en orden. Es más lento, pero confiable. Es el que usamos en el curso.
- **UDP** (*User Datagram Protocol*): sin conexión. Es más rápido, pero no garantiza la entrega ni el orden. Útil para streaming o juegos.

En la **arquitectura cliente-servidor**, el **servidor** ofrece un recurso o servicio y queda esperando, mientras los **clientes** se conectan a él para pedirlo.

## APIs, Webservices y Endpoints

Una **API** (Application Programming Interface) es un conjunto de reglas que define cómo un software puede comunicarse con otro. Cuando esta comunicación ocurre mediante **HTTP**, hablamos de una **API web** o **webservice**.

En términos simples: **una API web especifica cómo un cliente puede pedir datos a un servidor, cómo debe enviarlos y en qué formato se reciben las respuestas**.

Las APIs suelen organizar su información en **recursos**, cada uno identificado por un **path**. Ejemplos:
- `/usuarios`
- `/apuestas`
- `/casino/aviator`

Para interactuar con esos recursos se utilizan **métodos HTTP** como `GET`, `POST`, `PUT`, `PATCH` o `DELETE`. Estos le indican a la API qué tipo de consulta debe hacer sobre una url definida. En el curso las APIs se crean utilizando el módulo `flask`.

### Endpoints
Están definidos por:

**1. Método HTTP (la acción que quiero realizar)**
- `GET`: obtener datos.
- `POST`: crear un recurso nuevo.
- `PUT`: reemplazar **completamente** un recurso existente (sobrescribe todos los datos).
- `PATCH`: modificar **parcialmente** un recurso existente (solo actualiza los campos enviados).
- `DELETE`: eliminar un recurso.

**2. Path del recurso**
- Asociamos una función (código a ejecutar) cuando se hace una request a cada url.
- Ejemplo: `/usuarios/<id>`
- Puede incluir **parámetros de ruta** (por ejemplo, el ID del usuario).

**3. Parámetros de consulta (query params)**
Son opcionales y van después de `?`: `/usuarios?activo=true&limite=10`. Se pueden obtener con `flask.request.args.get("activo")` y `flask.request.args.get("limite")`.

**4. Cuerpo de la request (body)**
Generalmente se envía en **JSON**, indicando los datos a crear o modificar. Se puede obtener con `flask.request.get_json()`.

**5. Respuesta del servidor (response)**
La respuesta que entrega un servidor suele representarse mediante el objeto **`flask.Response()`**, el cual contiene:
- un **código de estado** (`status=`) HTTP (por ejemplo 200, 201, 400, 404, 500...)
- un **cuerpo** (`response=`), generalmente JSON.
- el **tipo de respuesta** (`content_type=`), generalmente `application/json`.

### Ejemplo de API
API con dos endpoints, con métodos `GET` y `POST` respectivamente.

`````python
from flask import Flask, request, Response, json

app = Flask(__name__)

usuarios = []
id_usuario = [0]


# GET: obtener un usuario
@app.route("/usuarios/<int:user_id>", methods=["GET"])
def obtener_usuario(user_id):
    for usuario in usuarios:
        if usuario["id"] == user_id:
            return Response(
                response=json.dumps(usuario),
                status=200,
                content_type="application/json",
            )


# POST: crear un usuario
@app.route("/usuarios", methods=["POST"])
def crear_usuario():
    body = request.get_json()  # Obtener JSON del body
    id_usuario[0] += 1

    nuevo_usuario = {
        "id": id_usuario[0],
        "nombre": body.get("nombre", "Sin nombre"),
    }
    usuarios.append(nuevo_usuario)

    return Response(
        response=json.dumps({"mensaje": f"Usuario {nuevo_usuario} creado con id: {id_usuario[0]}"}),
        status=201,  # Recurso creado
        content_type="application/json",
    )


if __name__ == "__main__":
    app.run(host="localhost", port=5001)
`````

# Serialización y Networking II 📦

## Pickle

- Serializa objetos a bytes, y no es legible para humanos.
- Puede serializar objetos de Python y clases personalizadas.
- Es específico para Python.
- Puede contener código malicioso.
- Métodos `dumps` y `dump` para serializar objetos. El segundo se usa para serializar directamente a un archivo.
- Métodos `loads` y `load` para deserializar objetos. El segundo se usa para deserializar directamente desde un archivo.

`````python
import pickle


class Alumno:

    def __init__(self, nombre, cursos, ayudantias):
        self.nombre = nombre
        self.cursos = cursos
        self.ayudantias = ayudantias

    def __str__(self):
        string = f"Alumno {self.nombre}\nCursos:\n"
        for curso in self.cursos:
            string += f"{curso}\n"
        string += f"Ayudantías:\n"
        for ayudantia in self.ayudantias:
            string += f"{ayudantia}\n"
        return string


alumno = Alumno("clemente",
                ["Diseño y Análisis de Algoritmos",
                 "Fundamentos Matemáticos para Ciencia de Datos",
                 "Algoritmos Paralelos en Computación Científica",
                 "Análisis de Regresión"],
                ["Programación Avanzada",
                 "Introducción a la Programación"])

print(alumno)
`````

`````python
# lo que nos entrega pickle no es muy legible
alumno_pickle = pickle.dumps(alumno)
print(alumno_pickle)
`````

`````python
# lo volvemos a cargar
alumno_2 = pickle.loads(alumno_pickle)
print(alumno_2)
`````

## Sockets

Un *socket* es un elemento que nos permite transmitir datos desde un dispositivo a otro. Un servidor y un cliente lo usan de la siguiente forma:

- Servidor
  - Crea el socket asociado al servidor.
  - Hace `bind` con su dirección IP y el puerto desde donde escuchará.
  - Hace `listen` para empezar a escuchar. Se le puede indicar una cantidad máxima de clientes conectados.
  - Hace `accept`, que queda esperando hasta que se conecte un nuevo cliente.
  - Hace `recv` y `send` con **el socket del cliente** para recibir y mandar información como bytes.
- Cliente
  - Crea el socket asociado al cliente.
  - Hace `connect` con la dirección IP y el puerto **del servidor**.
  - Hace `recv` y `send` con **su socket** para recibir y mandar información como bytes.

Para un servidor es fundamental el uso de *threads* en conjunto con networking para mantener la comunicación simultánea con todos los clientes. Asimismo, el cliente generalmente tiene un *thread* que se encarga de escuchar al servidor en todo momento.

- Arquitectura P2P: es un tipo de arquitectura alternativa a Cliente-Servidor. Cada usuario es a la vez cliente y servidor y se comunican de a pares (*peer to peer*). Es una arquitectura descentralizada ideal para cosas como transferencia de archivos.

El código de servidor y cliente que sigue es ilustrativo: cada uno se corre como un archivo aparte, ya que `accept` y `recv` bloquean la ejecución.

`````python
# servidor_tcp.py
import socket

# AF_INET: IPv4   |   SOCK_STREAM: TCP
servidor = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
servidor.bind(("localhost", 5000))
servidor.listen()

print("Esperando conexiones...")
socket_cliente, direccion = servidor.accept()   # bloquea hasta que llega un cliente

mensaje = socket_cliente.recv(1024)              # recibe bytes
print("Recibí:", mensaje.decode("utf-8"))

socket_cliente.send("Hola cliente".encode("utf-8"))
socket_cliente.close()
servidor.close()
`````

`````python
# cliente_tcp.py
import socket

cliente = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
cliente.connect(("localhost", 5000))             # se conecta al servidor

cliente.send("Hola servidor".encode("utf-8"))    # envía bytes
respuesta = cliente.recv(1024)
print("El servidor respondió:", respuesta.decode("utf-8"))

cliente.close()
`````

### Solución A

Un `Lock` permite que solo un thread a la vez entre a una sección crítica. `var_lock` debe ser un atributo de clase para que todos los escritores compartan el mismo lock. Si fuera un atributo de instancia, cada escritor tendría un lock distinto y no se protegería el acceso compartido al archivo.

`````python
import threading
from io import StringIO


class Escritor(threading.Thread):
    var_lock = threading.Lock()

    def __init__(self, archivo):
        super().__init__()
        self.archivo = archivo

    def run(self):
        with Escritor.var_lock:
            self.archivo.write(
                "Hola, soy un escritor escribiendo en un archivo de texto.\n"
            )


archivo = StringIO()
escritores = [Escritor(archivo) for _ in range(2)]

for escritor in escritores:
    escritor.start()
for escritor in escritores:
    escritor.join()

print(archivo.getvalue(), end="")
`````

### Solución D

El código original ejecuta las tareas secuencialmente porque espera a que termine `t1` antes de iniciar `t2`. `join()` bloquea el thread que lo llama hasta que termina el thread indicado. Para permitir concurrencia, primero se inician ambos threads y después se espera por ambos.

`````python
import threading
import time


def tarea(nombre, duracion):
    print(f"{nombre} comenzó")
    time.sleep(duracion)
    print(f"{nombre} terminó")


t1 = threading.Thread(target=tarea, args=("Tarea 1", 3))
t2 = threading.Thread(target=tarea, args=("Tarea 2", 3))

t1.start()
t2.start()
t1.join()
t2.join()
print("Programa terminado")
`````

### Solución: contadores

El `ServidorNorte` termina con exactamente `4_000_000`, porque el `Lock` protege cada incremento. El `ServidorSur` intenta alcanzar el mismo valor, pero al modificar el contador sin sincronización pueden ocurrir condiciones de carrera y perderse incrementos. Por eso su valor puede ser menor y no es confiable.

`````python
import threading

N_WORKERS = 4
N_MENSAJES = 1_000_000


class ServidorNorte:
    def __init__(self):
        self.contador = 0
        self.lock = threading.Lock()


class ServidorSur:
    def __init__(self):
        self.contador = 0


def trabajar_norte(servidor):
    for _ in range(N_MENSAJES):
        with servidor.lock:
            servidor.contador += 1


def trabajar_sur(servidor):
    for _ in range(N_MENSAJES):
        servidor.contador += 1


servidor_norte = ServidorNorte()
servidor_sur = ServidorSur()
threads = []

for _ in range(N_WORKERS):
    threads.append(threading.Thread(target=trabajar_norte, args=(servidor_norte,)))
    threads.append(threading.Thread(target=trabajar_sur, args=(servidor_sur,)))

for thread in threads:
    thread.start()
for thread in threads:
    thread.join()
print("ServidorNorte:", servidor_norte.contador)
print("ServidorSur:", servidor_sur.contador)
`````

``1) Encriptación de mensajes por chunks``

En una comunicación entre un cliente y un servidor, los mensajes no deberían enviarse directamente: primero deben transformarse a `bytes` y luego encriptarse. En este ejercicio implementarás el protocolo de comunicación utilizado por el sistema.

Tu misión es implementar las siguientes funciones:

```python
def encriptar(mensaje: bytes | bytearray) -> bytes | bytearray:
    ...


def desencriptar(mensaje: bytes | bytearray) -> bytes | bytearray:
    ...
```

### Requisitos

### Protocolo

1. Guarda el largo original del mensaje. Este valor será necesario para eliminar el relleno al desencriptar.
2. Divide el mensaje en segmentos ordenados de exactamente `124 bytes`, llamados **chunks**. Si el largo no es divisible por `124`, completa el último chunk concatenando los bytes `b"\\x00"` que sean necesarios.
3. Numera los chunks desde `0`. Cada número debe ocupar `4 bytes` y guardarse en formato `big endian`.
4. Forma cada paquete concatenando su identificador de 4 bytes con el chunk de 124 bytes. Por lo tanto, cada paquete tiene `128 bytes`.
5. Encripta cada paquete aplicando un desplazamiento simple a cada byte: suma `1` al valor del byte y utiliza el resto de la división por `256`.
6. El mensaje final debe comenzar con el largo original almacenado en `4 bytes` con formato `little endian`, seguido por todos los paquetes encriptados y ordenados.

La función `desencriptar` debe realizar el proceso inverso: leer el largo original, separar los paquetes de 128 bytes, restar `1` a cada byte (también usando módulo `256`), leer los identificadores en `big endian`, ordenar los chunks y eliminar el relleno usando el largo original.

Considera que se te entregara un mensaje valido de bytes o bytearray, y que el mensaje original puede ser vacío.




### Solución Propuesta


`````python
CHUNK_SIZE = 124
PACKET_SIZE = 4 + CHUNK_SIZE


def _desplazar(datos, cantidad):
    return bytes((byte + cantidad) % 256 for byte in datos)


def encriptar(mensaje: bytes | bytearray) -> bytes:
    mensaje = bytes(mensaje)
    paquetes = []

    for identificador, inicio in enumerate(range(0, len(mensaje), CHUNK_SIZE)):
        chunk = mensaje[inicio:inicio + CHUNK_SIZE]
        chunk += b"\x00" * (CHUNK_SIZE - len(chunk))
        numero = identificador.to_bytes(4, byteorder="big")
        paquetes.append(_desplazar(numero + chunk, 1))

    largo = len(mensaje).to_bytes(4, byteorder="little")
    return largo + b"".join(paquetes)


def desencriptar(mensaje: bytes | bytearray) -> bytes:
    mensaje = bytes(mensaje)
    if len(mensaje) < 4 or (len(mensaje) - 4) % PACKET_SIZE != 0:
        raise ValueError("El mensaje no tiene un formato válido")

    largo = int.from_bytes(mensaje[:4], byteorder="little")
    paquetes = {}

    for inicio in range(4, len(mensaje), PACKET_SIZE):
        paquete = _desplazar(mensaje[inicio:inicio + PACKET_SIZE], -1)
        identificador = int.from_bytes(paquete[:4], byteorder="big")
        if identificador in paquetes:
            raise ValueError("Hay identificadores de chunks repetidos")
        paquetes[identificador] = paquete[4:]

    esperado = (largo + CHUNK_SIZE - 1) // CHUNK_SIZE
    if set(paquetes) != set(range(esperado)):
        raise ValueError("Los identificadores de chunks no son consecutivos")

    datos = b"".join(paquetes[indice] for indice in sorted(paquetes))
    return datos[:largo]


mensajes = [b"", b"hola", b"a" * 124, bytearray(range(256))]

for original in mensajes:
    cifrado = encriptar(original)
    recuperado = desencriptar(cifrado)
    assert recuperado == bytes(original)

print("Todas las pruebas fueron exitosas.")
`````
