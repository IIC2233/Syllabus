| **Estructura** | **Tipo** | **Idea clave** | **Hoy** |
|:---------------|:---------|:---------------|:--------|
| `tuple` | Secuencial, **inmutable** | Elementos ordenados que no cambian. Al ser inmutable, es *hasheable* | — |

## 🧠 Resumen esencial de *stacks*

Un *stack* es **LIFO** (*Last In, First Out*): lo último que entra es lo primero en salir. En Python lo modelamos con una `list` normal, restringiéndonos a operar **solo en el final** de la lista.

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:----------------------|:------------|
| **Crear** | `stack = []` | Crea un *stack* vacío | `stack = []` |
| **Push** | `stack.append(elemento)` | Agrega un elemento al **tope** | `stack.append("uc.cl")` |
| **Pop** | `stack.pop()` | **Retorna y extrae** el elemento del tope | `sitio = stack.pop()` |
| **Peek** | `stack[-1]` | Retorna el tope **sin extraerlo** | `tope = stack[-1]` |
| **Largo** | `len(stack)` | Cantidad de elementos | `len(stack)` |
| **¿Vacío?** | `len(stack) == 0` o `not stack` | `True` si el *stack* está vacío | `if not stack: ...` |


```python
class DCCBrowser:

    def __init__(self):
        self.back_stack = []
        self.forward_stack = []
        self.sitio = "google.com"
    
    def empezar(self):
        while True:
            print(f"{'DCChrome':^62s}")
            print("-" * 62)
            print(f"{self.sitio:^62s}")
            print("-" * 62)
            print()

            sitio_atras = "-"
            sitio_adelante = "-"
            if self.back_stack:
                sitio_atras = self.back_stack[-1]
            if self.forward_stack:
                sitio_adelante = self.forward_stack[-1]
            
            print(f"1. Atrás: {sitio_atras}")
            print(f"2. Adelante: {sitio_adelante}")
            print("3. Nuevo sitio")
            print("4. Salir")
            nro = input("Elija una opción: ")
            
            if nro == "1":
                if not self.back_stack:
                    continue
                atras = self.back_stack.pop()
                self.forward_stack.append(self.sitio)
                self.sitio = atras
            
            elif nro == "2":
                if not self.forward_stack:
                    continue
                adelante = self.forward_stack.pop()
                self.back_stack.append(self.sitio)
                self.sitio = adelante
            
            elif nro == "3":
                self.forward_stack = []
                self.back_stack.append(self.sitio)
                self.sitio = input("Nombre del sitio: ")
            
            elif nro == "4":
                break

            else:
                print("Input incorrecto")

            

browser = DCCBrowser()
browser.empezar()
```

## Ejercicio 2: DCChrome II
En este ejercicio aprenderás a usar diccionarios. Los diccionarios son estructuras de datos en donde, en vez de asignarle una posición a cada valor, se le asigna una llave, con la cual se puede acceder al valor. Los valores pueden ser cualquier tipo de dato (incluyendo clases propias), mientras que las llaves solo pueden ser tipos de datos inmutables, por lo que no pueden ser listas, diccionarios, sets ni clases propias.

Programando tu navegador, te das cuenta que hay un detalle de seguridad crucial que no estás tomando en cuenta: las constraseñas. Para arreglar esto, añadirás una verificación adicional a la opción de "Nuevo sitio". En el `cuentas` tendrás un diccionario que contendrá nombres de sitios como llave, y el valor asociado será una tupla donde el primer elemento es el nombre de usuario y el segundo la contraseña asociada. Cuando el usuario introduzca un nuevo sitio, tendrás que tener en cuenta lo siguiente:

- Si el sitio está en el diccionario, deberás pedir el nombre de usuario y contraseña. Si estos coindicen al valor almacenado, se entra al sitio. En otro caso no se puede entrar y se vuelve a la pantalla de "Elegir opción".
- Si el sitio no está en el diccionario, deberás pedir al usuario que cree un nombre de usuario y contraseña. Deberás guardar este nuevo valor en el diccionario `cuentas`.

## 🔑 Resumen esencial de diccionarios

Un diccionario asocia una **llave** a un **valor**. La gracia es que acceder por llave toma **tiempo constante**: no importa cuántos elementos tenga el diccionario.

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:----------------------|:------------|
| **Crear** | `d = {}` / `d = {llave: valor}` | Crea un diccionario vacío o con pares iniciales | `{"google.com": ("juan", "1234")}` |
| **Acceder** | `d[llave]` | Retorna el valor asociado. **Lanza `KeyError`** si no existe | `d["google.com"]` |
| **Acceder seguro** | `d.get(llave, default)` | Igual que `[]`, pero retorna `default` (o `None`) si no existe | `d.get("uc.cl", None)` |
| **Insertar / actualizar** | `d[llave] = valor` | Crea la llave si no existe; la **sobrescribe** si ya existía | `d["uc.cl"] = ("ana", "abc")` |
| **Eliminar** | `del d[llave]` | Borra el par. También lanza `KeyError` si la llave no existe | `del d["uc.cl"]` |
| **¿Existe?** | `llave in d` | `True` si la **llave** está en el diccionario | `if "uc.cl" in d: ...` |
| **Recorrer llaves** | `d.keys()` | Todas las llaves | `for k in d.keys(): ...` |
| **Recorrer valores** | `d.values()` | Todos los valores | `for v in d.values(): ...` |
| **Recorrer pares** | `d.items()` | Tuplas `(llave, valor)` | `for k, v in d.items(): ...` |
| **Largo** | `len(d)` | Cantidad de pares | `len(d)` |
| **Por comprensión** | `{k: v for ... in ...}` | Construye un diccionario en una línea | `{p: len(p) for p in sitios}` |

> ⚠️ **Llaves permitidas:** solo objetos **hasheables**, es decir, los inmutables (`int`, `str`, `tuple`, `namedtuple`...). **No** se puede usar `list`, `dict` ni `set` como llave. Los **valores**, en cambio, pueden ser cualquier cosa.


```python
class DCCBrowser:

    def __init__(self):
        self.back_stack = []
        self.forward_stack = []
        self.sitio = "google.com"
        self.cuentas = {
            "google.com": ("juanitogugle", "juan123#"),
            "youtube.com": ("juanitoelpro", "juanyutu77!"),
            "chatgpt.com": ("juanitoflojin", "juanjuan_juan")
        }
    
    def empezar(self):
        while True:
            print(f"{'DCChrome':^62s}")
            print("-" * 62)
            print(f"{self.sitio:^62s}")
            print("-" * 62)
            print()

            sitio_atras = "-"
            sitio_adelante = "-"
            if self.back_stack:
                sitio_atras = self.back_stack[-1]
            if self.forward_stack:
                sitio_adelante = self.forward_stack[-1]
            
            print(f"1. Atrás: {sitio_atras}")
            print(f"2. Adelante: {sitio_adelante}")
            print("3. Nuevo sitio")
            print("4. Salir")
            nro = input("Elija una opción: ")

            if nro == "1":
                if not self.back_stack:
                    continue
                atras = self.back_stack.pop()
                self.forward_stack.append(self.sitio)
                self.sitio = atras
            
            elif nro == "2":
                if not self.forward_stack:
                    continue
                adelante = self.forward_stack.pop()
                self.back_stack.append(self.sitio)
                self.sitio = adelante
            
            elif nro == "3":
                sitio = input("Nombre del sitio: ")
                if sitio in self.cuentas.keys():
                    usuario = input("Nombre de usuario: ")
                    contrasena = input("Contrasena: ")
                    if self.cuentas[sitio] == (usuario, contrasena):
                        self.forward_stack = []
                        self.back_stack.append(self.sitio)
                        self.sitio = sitio
                    else:
                        print("Usuario o contraseña incorrecto")
                else:
                    usuario = input("Cree un nombre de usuario: ")
                    contrasena = input("Cree una contraseña: ")
                    self.cuentas[sitio] = (usuario, contrasena)
                    self.forward_stack = []
                    self.back_stack.append(self.sitio)
                    self.sitio = sitio
            
            elif nro == "4":
                break

            else:
                print("Input incorrecto")

            

browser = DCCBrowser()
browser.empezar()
print(browser.cuentas)
```

## Ejercicio 3: DCChrome III

En este ejercicio aprenderemos a usar _Named Tuples_. Las _Named Tuples_ son estructuras que nos permiten un funcionamiento similar a una clase, pero sin métodos, es decir, solo con atributos. Al igual que las tuplas, son inmutables y permiten acceso por índice.

Para darle un último toque al navegador, tendrás que crear una _named tuple_ llamada `Cuenta` que contendrá el nombre de usuario y la contraseña como atributos. Harás lo mismo que en el ejercicio anterior, pero en vez de usar tuplas para las cuentas, usarás la nueva _named tuple_ `Cuenta`.

## 🏷️ Resumen esencial de *named tuples*

Una *named tuple* es una tupla en la que cada posición tiene **nombre**. Sirve para agrupar datos relacionados sin escribir una clase completa: se lee `cuenta.usuario` en vez de `cuenta[0]`.

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:----------------------|:------------|
| **Importar** | `from collections import namedtuple` | Viene en el módulo `collections` | — |
| **Definir el tipo** | `Tipo = namedtuple("Nombre", [campos])` | Crea la "plantilla". Los campos van como lista de *strings* o un *string* separado por comas | `Cuenta = namedtuple("Cuenta", ["usuario", "contrasena"])` |
| **Instanciar** | `Tipo(valor1, valor2)` | Se crea como si fuera una función | `c = Cuenta("juanito", "1234")` |
| **Acceso por nombre** | `objeto.campo` | La forma legible y recomendada | `c.usuario` |
| **Acceso por índice** | `objeto[i]` | Sigue siendo una tupla, así que el índice funciona | `c[0]` |
| **Desempaquetar** | `a, b = objeto` | Igual que cualquier tupla | `usuario, clave = c` |
| **Copiar cambiando** | `objeto._replace(campo=valor)` | Retorna una **copia nueva** con ese campo cambiado | `c._replace(contrasena="5678")` |
| **Ver campos** | `Tipo._fields` | Tupla con los nombres de los campos | `Cuenta._fields` |

> ⚠️ **Son inmutables:** `c.usuario = "otro"` lanza un `AttributeError`. Para "modificar" hay que usar `_replace`, que en realidad crea otra *named tuple*. Como son inmutables, también son **hasheables** y sirven como llave de un diccionario.


```python
from collections import namedtuple

Cuenta = namedtuple("cuenta", ["usuario", "contrasena"])


class DCCBrowser:

    def __init__(self):
        self.back_stack = []
        self.forward_stack = []
        self.sitio = "google.com"
        self.cuentas = {
            "google.com": Cuenta("juanitogugle", "juan123#"),
            "youtube.com": Cuenta("juanitoelpro", "juanyutu77!"),
            "chatgpt.com": Cuenta("juanitoflojin", "juanjuan_juan")
        }
    
    def empezar(self):
        while True:
            print(f"{'DCChrome':^62s}")
            print("-" * 62)
            print(f"{self.sitio:^62s}")
            print("-" * 62)
            print()

            sitio_atras = "-"
            sitio_adelante = "-"
            if self.back_stack:
                sitio_atras = self.back_stack[-1]
            if self.forward_stack:
                sitio_adelante = self.forward_stack[-1]
            
            print(f"1. Atrás: {sitio_atras}")
            print(f"2. Adelante: {sitio_adelante}")
            print("3. Nuevo sitio")
            print("4. Salir")
            nro = input("Elija una opción: ")

            if nro == "1":
                if not self.back_stack:
                    continue
                atras = self.back_stack.pop()
                self.forward_stack.append(self.sitio)
                self.sitio = atras
            
            elif nro == "2":
                if not self.forward_stack:
                    continue
                adelante = self.forward_stack.pop()
                self.back_stack.append(self.sitio)
                self.sitio = adelante
            
            elif nro == "3":
                sitio = input("Nombre del sitio: ")
                if sitio in self.cuentas.keys():
                    usuario = input("Nombre de usuario: ")
                    contrasena = input("Contrasena: ")
                    if self.cuentas[sitio].usuario == usuario and self.cuentas[sitio].contrasena == contrasena:
                        self.forward_stack = []
                        self.back_stack.append(self.sitio)
                        self.sitio = sitio
                    else:
                        print("Usuario o contraseña incorrecto")
                else:
                    usuario = input("Cree un nombre de usuario: ")
                    contrasena = input("Cree una contraseña: ")
                    self.cuentas[sitio] = Cuenta(usuario, contrasena)
                    self.forward_stack = []
                    self.back_stack.append(self.sitio)
                    self.sitio = sitio
            
            elif nro == "4":
                break

            else:
                print("Input incorrecto")


browser = DCCBrowser()
browser.empezar()
print(browser.cuentas)
```
