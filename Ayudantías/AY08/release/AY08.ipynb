### Módulo `json` (Serialización y Deserialización)

|---|---|---|
| `json.dumps(obj, cls)` | Serializa un objeto Python a un *string* JSON.<br>**Argumentos:**<br>- `obj`: El objeto Python a serializar.<br>- `cls` (opcional): Subclase de `JSONEncoder` para aplicar serialización personalizada. | `json.dumps(m1, cls=MascotaEncoder)` |
| `json.loads(s, object_hook)` | Deserializa un *string* JSON a un objeto Python.<br>**Argumentos:**<br>- `s`: El *string* JSON de entrada.<br>- `object_hook` (opcional): Función que recibe un diccionario (resultado de cada objeto JSON) para personalizar la deserialización. | `json.loads(texto, object_hook=funcion_mascota_hook)` |

### Biblioteca `requests` (Cliente HTTP)

| Sintaxis / Clase / Método | Explicación | Ejemplo de Uso |
|---|---|---|
| `requests.get(url, timeout, params, headers)` | Realiza una solicitud HTTP mediante el método `GET`.<br>**Argumentos:**<br>- `url`: Ruta del recurso web.<br>- `timeout`: Define un límite de espera para evitar bloqueos indefinidos.<br>- `params`: Diccionario para codificar automáticamente el *query string*.<br>- `headers`: Diccionario con metadatos de la solicitud. | `requests.get(BASE, params=parametros, timeout=10)` |
| `requests.post(url, json, timeout, headers)` | Realiza una solicitud HTTP mediante el método `POST`, utilizado para procesar o crear recursos.<br>**Argumentos:**<br>- `json`: Diccionario que la biblioteca serializa y envía asignando `Content-Type: application/json`.<br>- `timeout`: Límite de espera.<br>- `headers`: Diccionario de cabeceras. | `requests.post(url, json=payload, timeout=10)` |
| `requests.put(url, json, timeout, headers)` | Realiza una solicitud HTTP mediante el método `PUT`, el cual crea o reemplaza el estado del recurso identificado.<br>**Argumentos:**<br>- `url`: Ruta del recurso web.<br>- `json`: Diccionario serializado en el cuerpo de la petición.<br>- `timeout`: Límite de espera de la solicitud.<br>- `headers`: Diccionario de cabeceras HTTP. | `requests.put(url, json=payload, timeout=10)` |
| `requests.patch(url, json, timeout, headers)` | Realiza una solicitud HTTP mediante el método `PATCH`, el cual aplica cambios parciales a un recurso.<br>**Argumentos:**<br>- `url`: Ruta del recurso web.<br>- `json`: Diccionario serializado con las modificaciones parciales.<br>- `timeout`: Límite de espera de la solicitud.<br>- `headers`: Diccionario de cabeceras HTTP. | `requests.patch(url, json=payload_parcial, timeout=10)` |
| `requests.delete(url, timeout, headers)` | Realiza una solicitud HTTP mediante el método `DELETE`, el cual solicita eliminar la asociación con el recurso.<br>**Argumentos:**<br>- `url`: Ruta del recurso web.<br>- `timeout`: Límite de espera de la solicitud.<br>- `headers`: Diccionario de cabeceras HTTP. | `requests.delete("http://localhost:4444/error/0")` |
| `respuesta.raise_for_status()` | Método de la respuesta que evalúa el código de estado y levanta una excepción si la solicitud resultó en un error (4xx o 5xx). | `respuesta.raise_for_status()` |
| `respuesta.status_code` | Atributo numérico que almacena el código de estado HTTP (ej. 200, 201, 404). | `print(respuesta.status_code)` |
| `respuesta.json()` | Método que interpreta y deserializa el cuerpo de la respuesta asumiendo formato JSON. | `data = respuesta.json()` |
| `respuesta.headers` | Diccionario con los campos de encabezado HTTP devueltos por el servidor. | `respuesta.headers.get("Content-Type", "")` |
| `respuesta.content` | Atributo que provee acceso a los datos binarios brutos (bytes) de la respuesta. | `len(respuesta_imagen.content)` |
| `respuesta.url` | Atributo que indica la URL codificada final gestionada por `requests`. | `print(respuesta.url)` |
| `respuesta.reason` | Atributo que provee la razón en formato texto asociada al código de estado de la respuesta. | `print(respuesta.reason)` |
| `respuesta.text` | Atributo que almacena el contenido del cuerpo de la respuesta en formato *string* decodificado. | `print(respuesta.text[:300])` |

### Framework Flask (Respuestas y Manejo de Errores)

| Sintaxis / Clase / Método | Explicación | Ejemplo de Uso |
|---|---|---|
| `return dict, status_code` | Retorno implícito de una vista. Flask convierte automáticamente un diccionario a JSON devolviendo código `200 OK` por defecto. Se puede especificar un código de error devolviendo una tupla. | `return {"error": "Faltan var_1 o var_2."}, 400` |
| `flask.Response(status, response, content_type)` | Instancia un objeto que permite controlar explícitamente todos los parámetros de la respuesta HTTP.<br>**Argumentos:**<br>- `status`: Código numérico de estado HTTP.<br>- `response`: Cuerpo con los datos de la respuesta.<br>- `content_type`: Cadena que describe el tipo MIME del contenido (ej. `text/plain`). | `return Response(status=200, response="Podemos mandar textos", content_type="text/plain; charset=utf-8")` |
| `flask.abort(codigo)` | Detiene la ejecución normal de la vista mediante una excepción `HTTPException`, devolviendo el código de error indicado.<br>**Argumentos:**<br>- `codigo`: Código numérico de estado HTTP correspondiente al error (ej. 400, 404). | `abort(404)` |

---

### Códigos de estado HTTP más comunes

| Código | Nombre | Cuándo se usa | Ejemplo en Flask |
|---|---|---|---|
| `200` | OK | La solicitud fue exitosa (valor por defecto). | `return datos` |
| `201` | Created | Se creó un recurso nuevo (típico de un `POST`). | `return nuevo, 201` |
| `204` | No Content | Éxito, pero sin cuerpo en la respuesta (típico de un `DELETE`). | `return "", 204` |
| `400` | Bad Request | La solicitud está mal formada: falta un campo o el cuerpo no es JSON válido. | `return {"error": "Faltan campos"}, 400` |
| `401` | Unauthorized | Faltan credenciales o no son válidas. | `return {"error": "No autorizado"}, 401` |
| `403` | Forbidden | Las credenciales son válidas, pero no tienen permiso para esa acción. | `return {"error": "Prohibido"}, 403` |
| `404` | Not Found | El recurso pedido no existe. | `return {"error": "No existe"}, 404` |
| `405` | Method Not Allowed | La ruta existe, pero no acepta ese método HTTP (Flask lo responde solo). | `@app.route("/juegos", methods=["GET"])` |
| `409` | Conflict | La solicitud choca con el estado actual (ej. un id repetido). | `return {"error": "Ya existe"}, 409` |
| `422` | Unprocessable Content | El formato es correcto, pero los valores no son válidos (ej. un número fuera de rango). | `return {"error": "Valor inválido"}, 422` |
| `500` | Internal Server Error | Error inesperado en el servidor (ej. una excepción no capturada). | — |

**Regla general:** `2xx` = éxito, `4xx` = error del cliente, `5xx` = error del servidor.
