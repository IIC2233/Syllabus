## Resumen de Programación Funcional 🧩

### 🔑 Paradigmas de programación

| **Paradigma** | **Idea clave** |
|:---------------|:----------------|
| Procedimental | Lista de instrucciones secuenciales que indican qué hacer paso a paso |
| Vectorial | Cálculos con paralelismo implícito; el compilador distribuye el trabajo |
| Declarativa | Se declara el problema y el computador decide cómo resolverlo (ej. SQL) |
| Orientada a Objetos | La solución se modela mediante objetos que interactúan entre sí |
| Funcional | La solución es un conjunto de funciones sin estado: el *output* depende solo del *input* |

> 💡 Python es **multiparadigma**: puedes combinar todos los enfoques anteriores en un mismo programa.

### ✅ Funciones puras

Una función es **pura** cuando cumple dos condiciones:

| **Criterio** | **Pregunta a hacerse** | **Lo que la rompe** |
|:--------------|:-------------------------|:----------------------|
| Determinismo | ¿El mismo *input* siempre entrega el mismo *output*? | Usar `random()` sin fijar una semilla propia dentro de la función |
| Sin efectos secundarios (*side effects*) | ¿La función deja todo igual fuera de su *scope*? | Modificar una lista/variable externa, hacer `print`, escribir un archivo |

> ⚠️ Un `print` dentro de una función ya la vuelve **impura**, aunque su valor de retorno sea determinista: altera algo fuera de su *scope* (lo que se muestra en pantalla).

> 💡 **¿Por qué importa?** Las funciones puras permiten memoización (`@cache`), son más fáciles de razonar —no hay que rastrear qué más cambia al llamarlas— y encajan naturalmente con flujos de transformación de datos (*data flow*).

### 🔁 Resumen esencial de funciones generadoras

| **Categoría** | **Comando / Sintaxis** | **Descripción breve** | **Ejemplo** |
|:--------------|:-----------------------|:------------------------|:--------------|
| **Definir** | función con `yield` en el cuerpo | Convierte la función en una **función generadora** | `def contar(): yield 1` |
| **Instanciar** | `gen = funcion_generadora(...)` | Al llamarla no se ejecuta nada todavía; retorna un objeto `generator` | `x = contar()` |
| **Avanzar** | `next(gen)` | Ejecuta hasta el próximo `yield` y retorna ese valor; recuerda dónde quedó | `next(x)` |
| **Recorrer** | `for x in gen:` | Llama a `next()` hasta que se agote (`StopIteration`) | `for i in contar(): ...` |
| **Enviar un valor** | `gen.send(valor)` | Reanuda el generador entregando `valor` a la expresión `yield` que lo pausó | `gen.send(5)` |
| **Delegar a otro iterable** | `yield from iterable` | Genera todos los valores de `iterable` sin necesidad de un `for` explícito | `yield from range(5)` |
| **¿Es generadora?** | Basta con que el cuerpo contenga un `yield` | No importa si ese `yield` se alcanza a ejecutar o no | `yield` dentro de un `if False:` sigue siendo generadora |

> 💡 **Ventaja principal:** un generador no guarda todos sus valores en memoria — los produce uno a uno, a medida que se piden. Por eso es ideal para leer archivos grandes línea por línea (`for linea in archivo`) en vez de `archivo.readlines()`.

## Resumen de Funciones sobre Iterables 🧮

### ⚙️ `map`, `filter`, `reduce` y `lambda`

| **Función** | **Firma** | **Qué hace** | **Ejemplo** |
|:-------------|:-----------|:----------------|:--------------|
| `map` | `map(f, iterable, ...)` | Aplica `f` a cada elemento y retorna un generador con los resultados. Con varios iterables, `f` debe recibir un parámetro por cada uno | `map(lambda x: x.lower(), strings)` |
| `filter` | `filter(f, iterable)` | Retorna un generador solo con los elementos donde `f(x)` es `True` | `filter(lambda x: x % 2 == 0, numeros)` |
| `reduce` | `reduce(f, iterable, inicial=None)` | Aplica `f(acumulado, siguiente)` sucesivamente hasta reducir el iterable a un solo valor | `reduce(lambda x, y: x + y, [1, 2, 3])` → `6` |
| `lambda` | `lambda parametros: expresion` | Función anónima de una sola expresión (sin `return`); ideal para usar directo dentro de `map`/`filter`/`reduce` | `lambda x, y: x + y` |

> ⚠️ **`reduce` con iterable vacío:** si no se entrega un valor `inicial` y el iterable está vacío, se lanza `TypeError`. Con un solo elemento y sin `inicial`, se retorna ese elemento sin aplicar `f`.

> ⚠️ **Orden importa:** en operaciones no conmutativas (como la división) o al reducir estructuras no ordenadas (como un `set`), el resultado de `reduce` puede depender del orden en que se recorran los elementos.

> 💡 **Funciones de primera clase:** en Python las funciones son objetos como cualquier otro — se pueden guardar en variables y pasar como argumento a otras funciones. Esto es justo lo que hace posible `map`, `filter` y `reduce`.

### 🧰 Otras funciones y librerías útiles

| **Categoría** | **Comando** | **Descripción breve** |
|:--------------|:-------------|:-------------------------|
| Índice + valor | `enumerate(iterable)` | Retorna tuplas `(indice, elemento)` |
| Combinar iterables | `zip(it1, it2, ...)` | Retorna tuplas con los elementos i-ésimos de cada iterable (se corta en el más corto) |
| Combinaciones | `itertools.combinations(iterable, r)` | Genera todas las combinaciones de largo `r`, sin reemplazo ni orden |
| Producto cartesiano | `itertools.product(it1, it2, ...)` | Genera todas las tuplas posibles combinando un elemento de cada iterable |
| Contar elementos | `sum(1 for _ in gen)` | Forma de obtener el largo de un generador (no tiene `__len__`) |
| ¿Todos/alguno cumple? | `all(map(f, it))` / `any(map(f, it))` | `True` si **todos**/**al menos uno** de los elementos cumplen `f` |
| Conteo de frecuencias | `collections.Counter(iterable)` | Diccionario que cuenta cuántas veces aparece cada elemento |

> 💡 **Consultas anidadas:** como `map`, `filter` y `reduce` siempre reciben y retornan iterables, se pueden encadenar directamente —por ejemplo `sum(map(f, filter(g, iterable)))`— para armar consultas complejas sin listas intermedias.
